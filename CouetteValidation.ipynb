```python
from pathlib import Path
import sys, subprocess, datetime, json
from IPython.display import Image, display
OUT = Path('outputs') / ('couette_' + datetime.datetime.now().strftime('%Y%m%d_%H%M%S'))
result = subprocess.run([sys.executable, 'validate_couette.py', '--out', str(OUT)], check=False)
print('Exit code:', result.returncode)
```

```python
s = json.loads((OUT / 'summary.json').read_text())
for row in s['rows']:
    print(row['case_id'], 'paper:', row['reference_spin'], 'PINN:', row['PINN_spin'], 'error %:', row['PINN_error_percent'])
print('All declared criteria:', s['all_declared_criteria_met'])
display(Image(filename=str(OUT / 'rotation_comparison.png')))
display(Image(filename=str(OUT / 'free_rotation.gif')))
```
