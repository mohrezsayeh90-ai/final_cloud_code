```python
from pathlib import Path
import sys, subprocess, datetime
from IPython.display import display, HTML, Image

PACK = Path('Validation_Pack')  # مسیر بستهٔ اصلی خودتان
PRESET = 'demo'               # smoke / demo / paper
OUT = Path('outputs') / datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
assert (PACK / 'case_catalog.json').is_file(), 'مسیر Validation_Pack را اصلاح کنید'
```

```python
result = subprocess.run([sys.executable, 'gaussian_parcel_pinn.py', '--preset', PRESET, '--validation-pack', str(PACK), '--out', str(OUT)], check=False)
print('Exit code:', result.returncode)
```

```python
import json
summary = json.loads((OUT / 'summary.json').read_text())
print('Verification:', summary['verification_passed'], '/', summary['verification_total'])
print('Literature validation:', summary['physical_validation_cases_run'])
print('Research criteria:', summary['research_quality_criteria_met'])
for c in summary['cases']:
    print(c['case'], c['residuals'], c['training']['termination'])
```

```python
display(Image(filename=str(OUT / 'figures' / 'architecture.png')))
for g in sorted((OUT / 'cases').glob('*/transport.gif')):
    print(g.parent.name)
    display(Image(filename=str(g)))
```
